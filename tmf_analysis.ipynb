#Step 7: Import + Load Data
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("tmf_data.csv")

df.head()

```python
# Step 8: Check Data
df.info()
df.isnull().sum()
```

```python
#Step 9: TMF Completeness
total_docs = len(df)
complete_docs = len(df[df['Status'] == 'Complete'])

completeness = (complete_docs / total_docs) * 100

print("Total:", total_docs)
print("Complete:", complete_docs)
print("Completeness %:", round(completeness, 2))
```

```python
#Step 10: Status Summary
status_summary = df['Status'].value_counts()

print(status_summary)
```

```python
# Step 11: Create Graph
status_summary.plot(kind='bar')

plt.title("TMF Document Status")
plt.xlabel("Status")
plt.ylabel("Count")

plt.show()
```

```python
#Step 12: Study-wise Analysis
study_summary = df.groupby('Study')['Status'].value_counts().unstack()

print(study_summary)
```
